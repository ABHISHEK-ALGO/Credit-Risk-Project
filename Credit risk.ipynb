# Feature Selection 
## For the Categorical Variables

since in chi square test null hypothesis is : the given categorical features are not related and
alternate hyothesis : they are related
now we got all values less than alpha means p-val <=alpha we reject null hypothesis we have sufficient evidence to reject null. hence we will take all thay all are related to output target column.

```python
# we are going to use chi square test to see whether categorical feature are related to our target variable i.e approved flag or not
from scipy.stats import chi2_contingency

for i in ['MARITALSTATUS','EDUCATION','GENDER','last_prod_enq2','first_prod_enq2']:
    chi2,pval,_,_= chi2_contingency(pd.crosstab(df[i],df['Approved_Flag']))
    print(i,'---',pval)
                                                
                                                
```

Now here we find that for all the categorical variables we have p value low then the alpha or significance level. Hence we can reject the null hypo (HO:which is they are not related) and we conclude that they are related to each other.

### Now on numerical columns

```python
# now working on numeric columns
numeric_columns=[]
for i in df.columns:
    if df[i].dtype != 'object' and i not in ['PROSPICTID','Approved_Flag']:
       numeric_columns.append(i)
```

```python
print(list(numeric_columns))
```

Now here in numerical variables there are more than 2 and we have other categorical variable Approved Flag as our targeted column .
we see there are two variables one is categorical (Approved Flag having 4 sub categories) and other is numerical  hence we use ANOVA TEST to find relationship between these.

```python
vif_data=df[numeric_columns]
total_columns = vif_data.shape[1]
columns_to_be_kept=[]
column_index=0
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
for i in range (0,total_columns):
    
    vif_value = variance_inflation_factor(vif_data, column_index)
    print (column_index,'---',vif_value)
    
    
    if vif_value <= 6:
        columns_to_be_kept.append( numeric_columns[i] )
        column_index = column_index+1
    
    else:
        vif_data = vif_data.drop([ numeric_columns[i] ] , axis=1)
```

```python
from scipy.stats import f_oneway

columns_to_be_kept_numerical = []

for i in columns_to_be_kept:
    a = list(df[i])  
    b = list(df['Approved_Flag'])  
    
    group_P1 = [value for value, group in zip(a, b) if group == 'P1']
    group_P2 = [value for value, group in zip(a, b) if group == 'P2']
    group_P3 = [value for value, group in zip(a, b) if group == 'P3']
    group_P4 = [value for value, group in zip(a, b) if group == 'P4']


    f_statistic, p_value = f_oneway(group_P1, group_P2, group_P3, group_P4)

    if p_value <= 0.05:
        columns_to_be_kept_numerical.append(i)
```

```python
# listing all the final features
features = columns_to_be_kept_numerical + ['MARITALSTATUS', 'EDUCATION', 'GENDER', 'last_prod_enq2', 'first_prod_enq2']
df = df[features + ['Approved_Flag']]
```
